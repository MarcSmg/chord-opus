```python
import re
from pathlib import Path
import json
```

```python
def merge_consecutive(lst):
    merged = [lst[0]] if lst else []
    merged.extend(curr for prev, curr in zip(lst, lst[1:]) if prev != curr)
    
    return merged

data = [1, 1, 2, 3, 3, 3, 4, 2, 2]
print(merge_consecutive(data))
```

```python
def compute_runs(lines):
    tonic = ''
    metre = ''
    runs = [dict()]
    
    for line in lines:
        if '# tonic' in line:
            tonic = line.split()[-1]
            
            if runs[-1].get('chords'): 
                runs.append(dict())

        if '# metre' in line:
            metre = line.split()[-1]
            if runs[-1].get('chords'): 
                runs.append(dict())
            
        if '|' in line:
            if not runs[-1].get('chords'):
                runs[-1]['metre'] = metre
                runs[-1]['tonic'] = tonic
                runs[-1]['chords'] = []
                
            split_line = line.split('|')[1:-1]
            chord_regex = r"[A-G][#b]?:" # regex pattern a chord should match
            
            for bar in split_line:
                bar = re.split(r'[ .]', bar)
                
                for el in bar:
                    if el.strip() in {'N', 'X', '&pause', '*'}:
                        runs.append({'tonic': tonic, 'metre': metre, 'chords': []})
                        
                    elif re.search(chord_regex, el):
                        runs[-1]['chords'].append(el)
                        
            runs[-1]['chords'] = merge_consecutive(runs[-1]['chords'])
            
    return [r for r in runs if r.get('chords')]
```

```python
folder_path = Path('../data/raw/mcgill-billboard-salami')

song_folders = [item.name for item in folder_path.iterdir() if item.is_dir()]

# print(len(song_folders))
```

```python
songs_list = []
for song_id in song_folders:
    song_dict = dict()
    with open(folder_path / f'{song_id}/salami_chords.txt', 'r') as song:
        lines = song.readlines()
    song_dict['song'] = str(song_id)
    song_dict["runs"] = compute_runs(lines)
    songs_list.append(song_dict)
songs_list.sort(key=lambda x: [int(x['song'])])
print(json.dumps(song_dict, indent=4))
```
